Visualize webcam

```python
webcam = cv2.VideoCapture(1)#put number of you webcam in case you have many webcam

while True:
    ret, frame = webcam.read()
    if not ret:
        break

    cv2.imshow("Webcam", frame)
    if cv2.waitKey(40) & 0xFF == ord('q'): #press q to exit
        break

webcam.release()
cv2.destroyAllWindows()
```
